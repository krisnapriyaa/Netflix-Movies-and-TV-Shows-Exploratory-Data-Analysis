Data Visualization

Visualization 1 – Movies vs TV Shows

```python
content_counts = df['type'].value_counts()
plt.figure(figsize=(8, 5))
plt.bar(content_counts.index,content_counts.values)
plt.title('Distribution of Movies and TV Shows')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')
plt.show()
```

Visualization 2 – Content released over the years

```python
year_counts = df['release_year'].value_counts().sort_index()
plt.figure(figsize=(12, 5))
plt.plot(year_counts.index,year_counts.values,marker='o')
plt.title('Netflix Content Released Over the Years')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(True)
plt.show()
```

Top 10 Content-Producing Countries

```python
# Remove missing values and split multiple countries
countries = df['country'].dropna().str.split(',')

# Convert country lists into individual rows
countries = countries.explode().str.strip()

# Count the number of titles for each country
country_counts = countries.value_counts().head(10)

# Create horizontal bar chart
plt.figure(figsize=(12, 6))
plt.barh(country_counts.index[::-1],
         country_counts.values[::-1])
plt.title('Top 10 Countries Producing Netflix Content')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

Movie Duration Distribution

```python
# Select only movies
movies = df[df['type'] == 'Movie'].copy()

# Extract numerical duration
movies['duration_num'] = (
    movies['duration'].str.extract(r'(\d+)').astype(float)
)

# Remove missing duration values
movie_duration = movies['duration_num'].dropna()

# Create histogram
plt.figure(figsize=(10, 5))
plt.hist(movie_duration, bins=20, edgecolor='black')
plt.title('Distribution of Movie Durations on Netflix')
plt.xlabel('Movie Duration (Minutes)')
plt.ylabel('Number of Movies')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

Calculate duration statistics

```python
print("Movie Duration Statistics")
print("Mean:", movie_duration.mean())
print("Median:", movie_duration.median())
print("Minimum:", movie_duration.min())
print("Maximum:", movie_duration.max())
print("Standard Deviation:", movie_duration.std())
```

Content Rating Distribution

```python
# Remove missing ratings
ratings = df['rating'].dropna()

# Remove incorrect duration values from rating column
ratings = ratings[~ratings.str.contains(
    r'^\d+\s*min$', case=False, regex=True
)]

# Count the top 10 ratings
rating_counts = ratings.value_counts().head(10)

# Create bar chart
plt.figure(figsize=(10, 5))
plt.bar(rating_counts.index, rating_counts.values)
plt.title('Top 10 Content Ratings on Netflix')
plt.xlabel('Content Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Print the rating counts

```python
print("Top 10 Netflix Content Ratings:")
print(rating_counts)
```

Correlation Heatmap

```python
# Create a copy
analysis_df = df.copy()

# Extract movie duration in minutes
analysis_df['movie_minutes'] = pd.to_numeric(
    analysis_df.loc[
        analysis_df['type'] == 'Movie', 'duration'
    ].str.extract(r'(\d+)')[0],
    errors='coerce'
)

# Extract TV show seasons
analysis_df['seasons'] = pd.to_numeric(
    analysis_df.loc[
        analysis_df['type'] == 'TV Show', 'duration'
    ].str.extract(r'(\d+)')[0],
    errors='coerce'
)

# Select numerical columns
numeric_data = analysis_df[
    ['release_year', 'movie_minutes', 'seasons']
]
print(numeric_data.describe())
```

Create the heatmap

```python
plt.figure(figsize=(8, 5))
correlation = numeric_data.corr()
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)
plt.title('Correlation Heatmap of Netflix Numerical Features')
plt.tight_layout()
plt.show()
```
